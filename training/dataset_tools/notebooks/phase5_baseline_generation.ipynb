```python
import torch
from diffusers import StableDiffusionXLPipeline

pipe = StableDiffusionXLPipeline.from_pretrained(
    "stabilityai/stable-diffusion-xl-base-1.0",
    torch_dtype=torch.float16,
    variant="fp16",
    use_safetensors=True
)

pipe = pipe.to("cuda")
```

```python
import os

output_dir = "/content/mentivor_baseline_outputs"
os.makedirs(output_dir, exist_ok=True)

benchmark_prompts = [
    "Mentivor standing front-facing, cinematic illustration, navy blue hoodie, simple black spiky hair, dot eyes",
    "Mentivor sitting on a couch, cinematic illustration style",
    "Mentivor walking outdoors, cinematic illustration style",
    "Mentivor close-up portrait, cinematic illustration style",
    "Mentivor holding an object, cinematic illustration style",
    "Mentivor under nighttime lighting, cinematic illustration style",
    "Mentivor in bright daylight, cinematic illustration style",
    "Mentivor with another character, cinematic illustration style",
]

seed = 42
for i, prompt in enumerate(benchmark_prompts, start=1):
    generator = torch.Generator("cuda").manual_seed(seed)
    image = pipe(prompt=prompt, num_inference_steps=30, guidance_scale=7.0, generator=generator).images[0]
    filename = f"{output_dir}/baseline_{i:02d}.png"
    image.save(filename)
    print(f"Saved: {filename}  |  prompt: {prompt}")
```
